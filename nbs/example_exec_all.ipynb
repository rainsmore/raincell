**The Answer: nbdev does NOT re-execute most cells during documentation generation.**

**"We don't, however, execute any other cells. That's because you wouldn't want to wait for your entire notebook to run just to build your docs; for instance, your docs might demonstrate training a model which takes hours to complete!"**

- **Cells with imports, `show_doc`, or exports**: get re-executed

Based on my research, here are the key findings and strategies for handling large map outputs with nbdev:

## Key Finding
**nbdev does NOT re-execute most cells during docs generation.** It only runs:
- Cells with `import` statements
- Cells with `show_doc`
- Exported cells

All other cells use their saved outputs. So **you DO need to keep outputs in the repo** for maps to appear in docs.

## Strategies for Large Outputs

1. **Selective Output Clearing with `nbdev_clean --clear_all`**
   - You can run `nbdev_clean --clear_all` on specific notebooks to strip outputs before committing
   - But then those outputs won't appear in docs

2. **Use Directives to Control What Appears**
   - `#| hide` - hides both code AND output from docs (useful for large intermediate outputs)
   - `#| output: false` - hides output but keeps code visible
   - `#| exec_doc` - forces re-execution during docs build (if you want to regenerate maps in CI)

3. **Save Maps Externally**
   - Save map images to a separate directory (e.g., `maps/`)
   - Commit only small placeholder/sample maps
   - Generate full maps during CI for docs deployment
   - Use relative links in notebooks: `![Map](maps/sample.png)`

4. **Git LFS for Large Files**
   - Store large notebook outputs with Git Large File Storage
